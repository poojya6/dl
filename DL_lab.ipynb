```python
#1
import numpy as np

# Vector operations
A = np.array([1, 2, 3])
B = np.array([4, 5, 6])

print("Vector Addition:", A + B)
print("Vector Subtraction:", A - B)
print("Dot Product:", np.dot(A, B))

# Matrix operations
M = np.array([[1, 2], [3, 4]])
N = np.array([[5, 6], [7, 8]])

print("Matrix Addition:\n", M + N)
print("Matrix Subtraction:\n", M - N)
print("Matrix Multiplication:\n", np.dot(M, N))
print("Transpose:\n", M.T)
```

```python
#2
import numpy as np

# Sample data
data = np.array([
    [1, 2],
    [2, 4],
    [3, 6]
])

# Calculate covariance matrix
cov = np.cov(data, rowvar=False)
print("Covariance Matrix:")
print(cov)

# Calculate eigenvalues and eigenvectors
values, vectors = np.linalg.eig(cov)

print("Eigenvalues:")
print(values)

print("Eigenvectors:")
print(vectors)
```

```python
#3
import numpy as np
from scipy.stats import binom, norm

# Binomial Distribution
n = 5
p = 0.5
result = binom.pmf(3, n, p)
print("Binomial Probability:", result)

# Normal Distribution
x = 70
mean = 60
std = 10
result = norm.pdf(x, mean, std)
print("Normal Distribution:", result)

# Bayes' Theorem
PA = 0.1
PBA = 0.9
PB_notA = 0.2

PB = (PBA * PA) + (PB_notA * (1 - PA))
PAB = (PBA * PA) / PB

print("Bayes Probability:", PAB)
```

```python
#4
import numpy as np

# Actual probability distribution
P = np.array([0.8, 0.2])

# Predicted probability distribution
Q = np.array([0.6, 0.4])

# Entropy
entropy = -np.sum(P * np.log(P))

# Cross Entropy
cross_entropy = -np.sum(P * np.log(Q))

# KL Divergence
kl_divergence = np.sum(P * np.log(P / Q))

print("Entropy:", entropy)
print("Cross Entropy:", cross_entropy)
print("KL Divergence:", kl_divergence)
```

```python
#5
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import entropy

# Actual probability distribution
P = np.array([0.8, 0.2])

# Predicted probability distribution
Q = np.array([0.6, 0.4])

# Calculate Entropy
H = entropy(P)

# Calculate Cross Entropy
CE = -np.sum(P * np.log(Q))

# Calculate KL Divergence
KL = entropy(P, Q)

print("Entropy:", H)
print("Cross Entropy:", CE)
print("KL Divergence:", KL)

# Visualization
labels = ['Class 1', 'Class 2']
x = np.arange(len(labels))
width = 0.35

plt.bar(x - width/2, P, width, label='Actual')
plt.bar(x + width/2, Q, width, label='Predicted')

plt.xlabel("Classes")
plt.ylabel("Probability")
plt.title("Actual vs Predicted Probability")
plt.xticks(x, labels)
plt.legend()
plt.show()
```
